```python
scores = [88, 72, 95, 61, 79, 84]
ratings = [4.2, 3.9, 5.0, 4.8, 4.1]

print(scores)
print(ratings)
```

Why is it helpful to test functions with small lists first?
So that the code can run with small data sets first to debug before it gets to bigger ones.

```python
def calculate_average(values):
    total = sum(values)
    count = len(values)
    return total / count

print(calculate_average(scores))
print(calculate_average(ratings))
```

Why should this function return the average instead of only printing it?
The function needs to be called to work with the data to get the average instead of simply printing.

```python
def count_above(values, threshold):
    count = 0
    for value in values:
        if value > threshold:
            count += 1
    return count

print(count_above(scores, 80))
print(count_above(ratings, 4.5))
```

How does the threshold parameter make this function reusable?
The threshold parameter makes this function resuable as it will accept any cutoff value instead on a single value.

```python
def count_passing(scores, passing_score):
    count = 0
    for score in scores:
        if score >= passing_score:
            count += 1
    return count

print(count_passing(scores, 70))
```

 What would change if the passing score were 80 instead of 70?
 Only the scores above 80 would be counted and then that would give a small count.

```python
def get_range(values):
    highest = max(values)
    lowest = min(values)
    return highest - lowest

print(get_range(scores))
print(get_range(ratings))
```

 What does the range tell you about a dataset?
 It shows how wide of data range one has for the scores and ratings.

```python
def summarize_scores(scores):
    average = calculate_average(scores)
    score_range = get_range(scores)
    passing_count = count_passing(scores, 70)

    return average, score_range, passing_count

avg, spread, passed = summarize_scores(scores)
print("Average:", avg)
print("Range:", spread)
print("Passed:", passed)
```

What is one advantage of having summarize_scores call smaller functions?
This helps the coding to leave the smaller functions distinct and intact and can make it easier to debug.

```python
test_scores = [80, 90, 100]

print("Expected average: 90")
print("Actual average:", calculate_average(test_scores))

print("Expected passing count: 3")
print("Actual passing count:", count_passing(test_scores, 70))
```

Why should test cases include expected results?
Expected results in test cases allow one to actually verify if the code is correct

```python
def count_above(values, threshold):
    count = 0
    for value in values:
        if value > threshold:
            count += 1
    return count

print(count_above([2, 5, 8], 4))
```
