You want to make sure the values match type, like when users type in a quiz score you want it to be a float.

```python
# Coding Task: score + attendance classifier
score = float(input("Enter quiz score: "))
attendance = float(input("Enter attendance rate (0-1): "))

if score >= 90 and attendance >= 0.90:
    label = "Excellent and eligible"
elif score >= 70 and attendance >= 0.80:
    label = "Pass and eligible"
else:
    label = "Needs review"

print("Score:", score)
print("Attendance:", attendance)
print("Label:", label)
```
