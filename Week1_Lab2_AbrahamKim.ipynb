```python
scores = [85, 92, 78, 90, 88, 73, 95]
print(scores)
```

One value represents 1 score

```python
print("First score:", scores[0])
print("Second score:", scores[1])
print("Last score:", scores[-1])
```

[0] is the first item in a list.

```python
score_count = len(scores)
score_total = sum(scores)

print("Number of scores:", score_count)
print("Total points:", score_total)
```

len(scores) returns number of scores

```python
highest_score = max(scores)
lowest_score = min(scores)

print("Highest score:", highest_score)
print("Lowest score:", lowest_score)
```

min() and max() show the highest and lowest values, which are easy to get from a list.

```python
average_score = score_total / score_count
print("Average score:", average_score)
```

Average score is totaling all of the scores and then dividing by the number of scores.

```python
print("Score Summary")
print("-------------")
print("Count:", score_count)
print("Total:", score_total)
print("Highest:", highest_score)
print("Lowest:", lowest_score)
print("Average:", average_score)
```

```python
weekly_study_hours = [3.5, 4, 2, 5, 6, 4.5, 3]

count = len(weekly_study_hours)
total_hours = sum(weekly_study_hours)
average = sum(weekly_study_hours) / count

print("Study Hours Summary")
print("Count:", count)
print("Total hours:", total_hours)
print("Highest:", max(weekly_study_hours))
print("Lowest:", min(weekly_study_hours))
print("Average:", average)
```
