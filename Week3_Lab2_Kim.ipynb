```python
student = {
    "name": "Maya",
    "score": 88,
    "hours_studied": 5,
    "attendance_rate": 0.95
}

print(student)
```

Why is a dictionary better than a list for representing this one student record?
A dictionary has more features like taking attributes from a set of data than a list.

```python
print(student["name"])
print(student["score"])
print(student["hours_studied"])
```

What happens if you misspell a key name?
It makes a key error

```python
student["status"] = "Pass"
student["score"] = 90

print(student)
```

What is the difference between adding a new key and updating an existing key?
Both use the same syntax and if a key doesn't exist Python add it as a new entry.

```python
students = [
    {"name": "Maya", "score": 88, "hours": 5},
    {"name": "Alex", "score": 92, "hours": 7},
    {"name": "Sam", "score": 75, "hours": 3},
    {"name": "Jordan", "score": 64, "hours": 2}
]

print(students)
```

In this structure, what does each dictionary represent?
Each dicitonary is one student's record, which has all the other data, score, hours, etc.

```python
for student in students:
    print(student["name"], student["score"])
```

Why is it useful that every record uses the same keys?
Consistent keys let you loop and access the same info every time.

```python
import pandas as pd

df = pd.DataFrame(students)
print(df)
```

What changed when the records became a DataFrame?
Pandas made a structured table with rows and columns.

```python
print("Shape:", df.shape)
print("Columns:", list(df.columns))

df.head()
```

5 rows and 4 columns
Column names: name, score, hours

```python
df.info()

df.describe()
```

info() shows the data type for each column and it becomes more specific.

```python
print(df["score"])

student_view = df[["name", "score"]]
print(student_view)

average_score = df["score"].mean()
print("Average score:", average_score)
```

Because inner brackets create a list of column names and passing list to df[...] tells pandas to return a dataframe with many columns.

```python
def label_score(score):
    if score >= 70:
        return "Pass"
    return "Needs Review"

df["status"] = df["score"].apply(label_score)
print(df)
```

 How does apply() connect a Python function to a pandas column?

apply() runs your function once for each value in the column, passing that value in the argument and passing each value into a new Series

```python

import pandas as pd

products = [
    {"product": "Keyboard", "category": "Accessories", "quantity": 3, "price": 49.99, "rating": 4.5},
    {"product": "Mouse",    "category": "Accessories", "quantity": 5, "price": 19.99, "rating": 4.2},
    {"product": "Monitor",  "category": "Displays",    "quantity": 2, "price": 149.99, "rating": 4.7},
    {"product": "Laptop",   "category": "Computers",   "quantity": 4, "price": 899.99, "rating": 4.8},
    {"product": "Webcam",   "category": "Accessories", "quantity": 6, "price": 39.99, "rating": 3.9}
]

df = pd.DataFrame(products)

print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

print(df.head())

df["total_value"] = df["quantity"] * df["price"]

average_rating = df["rating"].mean()
print("Average rating:", round(average_rating, 2))
```
